```python
import pandas as pd
```

```python
data= {'Students': ['Ahmad','Qaiyim','Intan','Fatehah','Akmal'], 'Marks': [85,None,45,None,67]}
```

```python
df=pd.DataFrame(data)
print('Original Data')
df
```

```python
print('Missing Data')
print('Number of missing data: ', df.isnull().sum())
```

```python
avg_marks=df['Marks'].mean()
print(avg_marks)
```

```python
df.fillna(avg_marks, inplace=True)
```

```python
print('after fillna() call()')
df
```

# Dropna() example

```python
emp_data={'Name':['Ahmad','Qaiyim','Intan','Fatehah','Khadijah','Sofea'],
         'Age':[26,None,25,10,34,None],
         'Salary':[5500,None,2400,4000,None,4500]}

df=pd.DataFrame(emp_data)
print('Original Data')
df
```

```python
print(df.isnull().sum())
```

```python
df_drop=df.dropna()
df_drop
```

```python
df_drop_salary=df.dropna(subset=['Salary'])
df_drop_salary
```

```python
df.fillna({'Age': df['Age'].mean(),'Salary': df['Salary'].mean()}, inplace=True)
df
```

```python
df=pd.read_csv('students_performance_dirty.csv')
print(df.head(5))

# Solution for Question 1
print('\nMissing Data')
print('Number of missing data: ', df.isnull().sum())

#SOlution for Question 2
df=df.dropna(subset=['gender'])
print('\nNumber of missing data: ', df.isnull().sum())

avg_study_hours=df['study_hours'].mean()
df.fillna(avg_study_hours, inplace=True)
print('\nNumber of missing data: ', df.isnull().sum())


```
