**Matrix Multiplication**
- tf.matmul() method
** Rules**
1. Inner dimension must match
    (m,n)@(n,p)
    for e.g
    (5,7)@(8,9) It will not work because n are different
    (5,8)@(8,9) It will work because n are same
  
2. The result matrix has the shape of outer dimension means:
    A (m,n)
    B (n,p)
    SO result has (m,p)
    e.g : (4,7)@(7,5) --> (4,5)

```python
import tensorflow as tf
import numpy as np
```

```python
tensor_A = tf.constant([[1,2],[3,4],[4,5]], dtype=tf.int32)
tensor_B = tf.constant([[5,6],[7,8]], dtype=tf.int32)
print('Shape of A',tensor_A.shape)
print('Shape of B',tensor_B.shape)
```

```python
tensor_C = tf.matmul(tensor_A,tensor_B)
print(tensor_C)
print('Shape of tensor_C',tensor_C.shape)
```

```python
X = tf.constant([
    [1.,0.,-1.],
    [2.,3.,1.]
],dtype=tf.float32)
X
```

```python
y = tf.constant([
    [4.],
    [5.],
    [6.]
], dtype=tf.float32)
y
```

```python
print('X shape:',X.shape)
print('Y shape:',y.shape)
```

```python
Z = tf.matmul(X,y)
Z
```

```python
print(f'X-shape: {X.shape}, Y-shape: {y.shape}, Z-shape{Z.shape}')
print(X.numpy())
print(y.numpy())
print(Z.numpy())
```

**Transpose:**

Transpose swaps axes. For 2D matrix shape (m,n) become (n,m) after transpose

```python
tensor_one = tf.constant([
    [[1,2],[2,3]],
    [[4,5],[6,7]]
],dtype=tf.int32)
print(tensor_one)
print('\nShape of tensor one: ',tensor_one.shape)
```

```python
tensor_one_transpose = tf.transpose(tensor_one)
print(tensor_one_transpose)
print('\nShape of tensor_one_transpose: ',tensor_one_transpose.shape)
```

3D Transpose with explicit axis permutation

```python
tensor_two = tf.constant([
    [[1,2],[3,4]],
    [[5,6],[7,8]]
],dtype=tf.int32)
print(tensor_two)
```

```python
# swap first and last (0,1,2) --> (0,2,1)
tensor_two_transpose = tf.transpose(tensor_two,perm=[0,2,1])
print(tensor_two_transpose)
print('\nShape of tensor_two_transpose: ',tensor_two_transpose.shape)
print('\nShape of tensor_two: ',tensor_two.shape)
```

**Reshape**

tf.reshape(tensor,new_shape)

Change the shape without change the data

```python
tensor= tf.constant([
    [1,2,3],
    [4,5,6]
],dtype=tf.int32)
print(tensor)
print('Shape of tensor: ',tensor.shape)
print('Number of elements are: ',tf.size(tensor).numpy())
```

```python
tensor_n1 = tf.reshape(tensor,shape=(3,2))
print(tensor_n1)
print('Shape of tensor: ',tensor_n1.shape)
print('Number of elements are: ',tf.size(tensor_n1).numpy())
```

```python
tensor_n2 = tf.reshape(tensor,shape=(6,1))
print(tensor_n2)
print('Shape of tensor: ',tensor_n2.shape)
print('Number of elements are: ',tf.size(tensor_n2).numpy())
```

```python
tensor_n3 = tf.reshape(tensor,shape=(-1,2))
print(tensor_n3)
print('Shape of tensor: ',tensor_n3.shape)
print('Number of elements are: ',tf.size(tensor_n3).numpy())
```

```python
tensor_n2 = tf.reshape(tensor,shape=(-1,6))
print(tensor_n2)
print('Shape of tensor: ',tensor_n2.shape)
print('Number of elements are: ',tf.size(tensor_n2).numpy())
```
