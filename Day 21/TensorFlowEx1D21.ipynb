**Arthmetic Operations with tensors**
- Add
- Sub
- Multinomial
- Div

```python
tensor_one = tf.constant([[11,9],[2,5]])
tensor_one
```

```python
# tensor_one+15
tf.add(tensor_one,15) # second way
```

```python
# tensor_one-4
tf.subtract(tensor_one,4)
```

```python
# tensor_one*3
tf.multiply(tensor_one,3)
```

```python
# tensor_one/3
tf.divide(tensor_one,3)
```

**Exercise**
Create two tensors with random values between 0 and 1, shape is [6,300]
- Number of elements
- Size,
- Rank,
- Shape

```python
import numpy as np

# Two tensors with random values between 0 and 1
tensor1 = np.random.rand(6, 300)
tensor2 = np.random.rand(6, 300)
```

```python
print("Tensor 1:")
print("Shape:", tensor1.shape)
print("Number of elements:", tensor1.size)
print("Rank (number of dimensions):", tensor1.ndim)
print("Data type:", tensor1.dtype)

print("\nTensor 2:")
print("Shape:", tensor2.shape)
print("Number of elements:", tensor2.size)
print("Rank (number of dimensions):", tensor2.ndim)
print("Data type:", tensor2.dtype)
```
