# KNN Algorithm in Supervised Machine Learning
# KNeighborClassifier of sklearn library

```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier 
```

```python
# Data Collection
X = np.array([
    [1,3500],
    [2,3200],
    [3,3000],
    [4,2500],
    [5,2300],
    [6,2000],
    [7,2200],
    [2,3700],
])
y= ['Unfit','Unfit','Unfit','Fit','Fit','Fit','Fit','Unfit']
```

```python
# Create the model
model = KNeighborsClassifier(n_neighbors=3)
```

```python
# Train the model
model.fit(X,y)
```

```python
# Predict
new_person = [[4.5,2600]]
predict = model.predict(new_person)
print('Predicted Result:', predict[0])
```

```python
# Predict from users
ex_hours=float(input('Enter your exercise hours'))
calory_intake=float(input('Enter your calory consume'))
predict_fit=model.predict([[ex_hours,calory_intake]])
print(f'Exercise Hour {ex_hours}, calory intake {calory_intake}, You are likely {predict_fit[0]}')
```

# Second example

```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
```

```python
data=pd.read_csv('fruit_data.csv')
data
```

```python
# Split the Data
X = data[['Weight','Size']]
y = data['Fruit']
```

```python
# Model Creation
model = KNeighborsClassifier(n_neighbors=3)
```

```python
# Train the model
model.fit(X,y)
```

```python
weight=float(input('Enter the fruit weight'))
size=float(input('Enter the fruit size'))
our_fruit= pd.DataFrame([[weight,size]], columns=['Weight','Size'])
predict_fruit=model.predict(our_fruit)
print(f'When weight is {weight}, size is {size}, the fruits is {predict_fruit[0]}')
```

# Third Example

```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
```

```python
data=pd.read_csv('student_grade.csv')
data
```

```python
#Split the data
X = data[['Hours_Studied', 'Assignments_Completed']]
y = data['Grade']
```

```python
# Create the model
model = KNeighborsClassifier(n_neighbors=3)
```

```python
# Train the model
model.fit(X,y)
```

```python
new_hour=float(input('Enter the hour studied'))
new_assgn=float(input('Enter the amount of assignment that has been submitted'))
our_grade= pd.DataFrame([[new_hour,new_assgn]], columns=['Hours_Studied','Assignments_Completed'])
predict_grade=model.predict(our_grade)
print(f'When studied hours is {new_hour}, and number of submitted assignment is {new_assgn}, the grade is likely {predict_grade[0]}')
```
