```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder
import numpy as np
```

```python
# Collect the data
data=pd.read_csv('pass_fail.csv')
data.head()
```

```python
# Split the data
X = data[['Hours']]
y = data['Result']
```

```python
# Create the model
model = LogisticRegression()
```

```python
# Train the model
model.fit(X,y)
```

```python
hour_study=float(input('Enter your hours of study'))
our_result= pd.DataFrame([[hour_study]], columns=['Hours'])
predicted_value=model.predict(our_result)
print(f'After {hour_study} hours, you are likely to {predicted_value}')
```

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder
import numpy as np
```

```python
data=pd.read_csv('customer_purchase.csv')
data.head()
```

```python
# Split the data
le=LabelEncoder()
data['Purchased']= le.fit_transform(data['Purchased'])
X=data[['Age','Salary']]
y=data['Purchased']
```

```python
data
```

```python
# Crate the model
model=LogisticRegression()

# Train the model
model.fit(X,y)
```

```python
new_cust_age = int(input('Enter age'))
new_cust_sal = int(input('Enter your salary'))
our_cust= pd.DataFrame([[new_cust_age,new_cust_sal]], columns=['Age','Salary'])
prediction= model.predict(our_cust)
if (prediction):
    print('Purchased')
else:
    print('Not purchased')
```

```python
new_customer=[[28,35000]]
our_customer= pd.DataFrame([[28,3500]], columns=['Age','Salary'])
prediction = model.predict(our_customer)
print(f'Predicted Result (1=Yes,0=No): {prediction[0]}')
```

# Logistic Regression - Spam/NoSpam

## Do the data cleaning and labelling (if required)
## Use Logistic Regression Algorithm to train the model
## Predict Result if the email is spam or not spam


```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import CountVectorizer
```

```python
data=pd.read_csv('spam_dataset.csv')
data
```

```python
X = data['message_content']
y = data['is_spam']
```

```python
# Convert text to numbers
cv = CountVectorizer()
X_cv = cv.fit_transform(X)
```

```python
model = LogisticRegression()
```

```python
model.fit(X_cv,y)
```

```python
sample = ['Good Morning Qaiyim','We have shocking sales!']
predict = model.predict(cv.transform(sample))
if predict[0]:
    print('NOT SPAM')
else:
    print('SPAM')
```
