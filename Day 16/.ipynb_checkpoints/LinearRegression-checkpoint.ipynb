# Supervised Machine Learning
## Linear Regression

```python
# import required library
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python

data=pd.read_csv('sales_data.csv')
data.head()
```

```python
# Splitting the data
X = data[['Advertising_Spend']]
y = data['Sales_Revenue']
```

```python
# Model Creation
model = LinearRegression()
# Model Train
model.fit(X,y)
print('Training Done')
```

```python
# Prediction
new_spend=float(input('Enter new spend to predict sales revenue'))
predicted_revenue = model.predict([[new_spend]])
print(f'Predicted sales revenue for RM {new_spend} add spend is {predicted_revenue.round(2)}')
```

# Exercise House Price

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

data=pd.read_csv('house_price.csv')
data.head()

house_size= data[['Size_sqft']]
house_price= data['Price']

model = LinearRegression()
model.fit(house_size,house_price)

new_size=float(input('Enter new squarefit to predict the house price'))
predicted_price = model.predict([[new_size]])
print(f'Predicted house price for {new_size} squarefeet is RM {predicted_price.round(2)}')
```
