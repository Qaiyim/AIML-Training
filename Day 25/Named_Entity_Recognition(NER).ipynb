```python
import spacy
nlp = spacy.load("en_core_web_sm")
print('SpaCy loaded Successfully')
```

```python
text = 'Ahmad Qaiyim was born in Malaysia in Year 1999'
```

```python
document = nlp(text)
for ent in document.ents:
  print(f'Text: {ent.text},Label: {ent.label_},Explanation: {spacy.explain(ent.label_)}')
```

```python
from spacy import displacy
displacy.render(document,style='ent',jupyter=True)
```

```python
corpus=[
    'KLCC is located in Malaysia',
    'Google was founded by Larry Page',
    'Ronaldo is the best player in Portugal',
    'Discovery of Malaysia was written by Tun Mahathir'
]
```

```python
for text in corpus:
  document = nlp(text)
  for ent in document.ents:
    print(f'Text: {ent.text},Label: {ent.label_},Explanation: {spacy.explain(ent.label_)}')
```

```python
from spacy import displacy
for text in corpus:
  document = nlp(text)
  displacy.render(document,style='ent',jupyter=True)
```
