# NLP PIPELINE

Import required libraries

```python
import nltk
import spacy
import re
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4') # for multi language
nltk.download('punkt_tab')
```

```python
our_text = 'Artificial Intelligence and Machine Learning are going to change the future of Technology'
```

Splits text into individual word/tokens (Tokenization)

```python
tokens = word_tokenize(our_text)
tokens
```

Cleaning

```python
clean_tokens = [re.sub(r'[^a-zA-Z]','',token.lower()) for token in tokens if token.isalpha()]
clean_tokens
```

Stop word removal

```python
# Remove common less meaningful words
stop_words = set(stopwords.words('english'))
filtered_tokens = [token for token in clean_tokens if token not in stop_words]
filtered_tokens
```

```python
# Stemming
stemmer = PorterStemmer()
# Reduce words to their root word form
stemmed_tokens = [stemmer.stem(token) for token in filtered_tokens]
stemmed_tokens
```

```python
# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(token) for token in filtered_tokens]
# Convert words to their dictionary form (lemma) using POS: Part Of Speech
lemmatized_tokens
```

Bag of Words(BOW)

```python
from nltk import corpus
corpus = [
    'Artificial Intelligence is transferring industries',
    'Machine Learning is a branch of artificial intelligence',
    'Deep Learning is a subfield of machine learning',
    'Natural Language Processing is a subfield of artificial intelligence'

]
```

```python
cv = CountVectorizer()
bow_matrix = cv.fit_transform(corpus)
print('*****Features Names BOW*****')
print(cv.get_feature_names_out())
print('*****BOW Matrix Array*****')
print(bow_matrix.toarray())
```
