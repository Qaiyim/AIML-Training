**Resume Skills Summary NLP**

```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from numpy import vectorize
```

```python
df = pd.read_csv('/content/sample_data/UpdatedResumeDataSet.csv',encoding='utf-8')
print('Shape of Data Frame: ',df.shape)
df.head(10)
```

```python
print('\n Columns',df.columns.tolist())
```

```python
sample_df = df.sample(n=10,random_state=42).reset_index(drop=True)
sample_df
```

```python
sample_df = sample_df.rename(columns={'Resume':'Resume_Text'})
sample_df
```

```python
vectorizer = TfidfVectorizer(stop_words='english',max_features=5000)
tfidf_matrix = vectorizer.fit_transform(sample_df['Resume_Text'])
tfidf_matrix.shape
```

```python
similarity_matrix = cosine_similarity(tfidf_matrix)
sim_df = pd.DataFrame(similarity_matrix,index=sample_df['Category']+'_'+sample_df.index.astype(str),
                      columns=sample_df['Category']+'_'+sample_df.index.astype(str))
sim_df.round(2).head(3)
```

**Resume Screening Example 2**

```python
resume ={
    'Candidates':['Qaiyim','Ahmad','Intan'],
    'Resume':[
        'Python Developer skilled in Django,Flask, Rest API and MySQL',
        'Machine learnig engineer experience in Python,Tensorflow,NLP and Data Analysis',
        'Full stack developer with .net, javascript, React, Node.js and MongoD8 experience']
}
```

```python
df = pd.DataFrame(resume)
df
```

```python
vectrorizer = TfidfVectorizer(stop_words='english',max_features=5000)
tfidf_matrix = vectrorizer.fit_transform(df['Resume'])
tfidf_matrix.shape
tfidf_matrix
```

```python
similarity_matrix = cosine_similarity(tfidf_matrix)
sim_df = pd.DataFrame(similarity_matrix,index=df['Candidates'],columns=df['Candidates'])
sim_df.round(2)
```
