```python
# Step 1
import numpy as np
import random
```

```python
# Step 2 Define states and action
states = np.arange (0,5,1)
actions = ['Green','Red']
print('States', states)
print('Actions',actions)
```

```python
# Step 3: Initialize the Q-TAble and Hyperparameters
Q = np.zeros((len(states), len(actions)))

alpha = 0.1    # Learning rate
gamma = 0.9    # Discount rate
epsilon = 0.2  # Exploration rate
episodes = 300 # Training runs

print('Q-table shape: ', Q.shape)
```

```python
def get_reward(traffic, action):
    reward = 0
    
    if 3 <= traffic <= 4:
        if action == 'Green':
            reward = 10 # Good: Clear congestion
        else:
            reward = -10 # Bad: Create jams
        
    elif traffic == 0:
        if action == 'Red':
            reward = 5  # Good: Save energy
        elif action == 'Green':
            reward -=5   # Bad: Waste power

    if traffic == 1 or traffic ==2:
        reward == 1         # Neutral
        
    return reward

print('Reward example (Traffic=4, Green):', get_reward(4, 'Green'))
```

```python
def next_traffic(traffic, action):
    if action == 'Green':      
        traffic -= random.choice([0, 1]) # When light is GREEN, traffic usually decreases
    else:
        traffic += random.choice([0, 1]) # # When light is RED, traffic may increase 
    return int(np.clip(traffic, 0, 4))  # Keep traffic level within 0–4 range
 
print('Next Traffic Example:', next_traffic(3, 'Green'))
```

```python
for ep in range(episodes):
    traffic = random.choice(states)  # random starting state 

    for _ in range(15):  # Steps per episode
       # Choose action (exploration vs exploitation)
        if random.uniform(0, 1) < epsilon:
            action = random.choice(actions)
        else:
            action = actions[np.argmax(Q[traffic])]  

        # Take Action → Get Next State and reward
        next_state = next_traffic(traffic, action)
        reward = get_reward(traffic, action)

    
        a = actions.index(action)
        best_next = np.max(Q[next_state])  

        # Q-Learning update rule
        Q[traffic, a] = Q[traffic, a] + alpha * (reward + gamma * best_next - Q[traffic, a])

        # Move to next state
        traffic = next_state

print('Training Completed')
```

```python
try:
    traffic = int(input('Enter starting traffic level (0–4): '))
    if traffic < 0 or traffic > 4:
        raise ValueError('Traffic level out of range!')
except ValueError as e:
    print(e)
    traffic = 2
    print('Default level set to 2 (moderate traffic)')

print(f'\nStarting with traffic level : {traffic}')
print(f'Simulating 10 steps of smart traffic light control:\n')

for step in range(10):
    action = actions[np.argmax(Q[traffic])]
    print(f'Step {step+1}: Traffic Level={traffic} → Action={action}')
    traffic = next_traffic(traffic, action)

print('\nSimulation complete. Traffic light control finished!')
```
