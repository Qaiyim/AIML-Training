# Multi Label Classification example using Tensor Flow

```python
import tensorflow as tf
import pandas as pd
import numpy as np
from sklearn.preprocessing import MultiLabelBinarizer
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
data = {
    'overview':
    [
        'A Hero Save The City From Destruction',
        'A Girl And Boy Become Friends During War',
        'A Funny Story Of Collegues During Office Meet',
        'One Scary Night In A Deserted Place'
    ],
    'genres':[['Action','Adventure'],['Romance','Drama'],['Comedy'],['Horror']]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
mlb = MultiLabelBinarizer()
X = df['overview']
y = mlb.fit_transform(df['genres'])
```

```python
y
```

```python
tokenizer = Tokenizer(num_words=5000, oov_token='<OOV>')
tokenizer.fit_on_texts(X)
X = tokenizer.texts_to_sequences(X)
X = pad_sequences(X, maxlen=200,padding='post')
```

```python
# Model Creation
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(input_dim=5000, output_dim=16, input_length=200),
    tf.keras.layers.GlobalAveragePooling1D(),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(len(mlb.classes_), activation='sigmoid'),
])
```

```python
# Model compile
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
# Train the model
model.fit(X, y, epochs=20)
```

```python
history = model.fit(X, y, epochs=20)
```

```python
# Prediction on a new movie overview
new_overview = ['A lonely wolf was walking in forest']
new_overview = tokenizer.texts_to_sequences(new_overview)
new_overview = pad_sequences(new_overview, maxlen=20, padding='post')
prediction = model.predict(new_overview)
prediction_labels = [mlb.classes_[i] for i in range(len(prediction[0])) if prediction[0][i]>= 0.5]
print(prediction_labels)
```
