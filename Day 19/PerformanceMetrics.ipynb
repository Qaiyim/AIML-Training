# MAE: Mean Absolute Error

```python
from sklearn.metrics import mean_absolute_error
```

```python
actual_value = [70,80,90,85,95]
predicted_value = [68,82,88,89,97]
```

```python
mae = mean_absolute_error(actual_value, predicted_value)
print("Mean Absolute Error:", mae)
```

# MSE: Mean Squared Error

```python
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
```

```python
house_size = np.array([[1000],[1500],[2000],[2500],[3000]])
house_price = np.array([120000,180000,220000,280000,320000])
```

```python
model=LinearRegression()
```

```python
model.fit(house_size,house_price)
```

```python
price_predict = model.predict(house_size)
```

```python
mse = mean_squared_error(house_size,price_predict)
print("Mean Absolute Error:", mse)
```

```python
print('MSE for this model: ',mse)
```

```python
plt.figure(figsize=(8,5))
plt.scatter(house_size,house_price, color='maroon', label='Actual Value', s=100)
plt.plot(house_size,price_predict, color='blue', label='Predicted Value ', linewidth=2)
plt.legend()
```
