```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score,r2_score
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
data = {
    'StudyHours': [2, 3, 4, 5, 6, 7, 8, 9, 10],
    'Attendance': [60, 65, 70, 72, 75, 80, 85, 88, 90],
    'PrevScore': [40, 45, 50, 55, 60, 65, 70, 75, 80],
    'FinalMarks': [50, 55, 60, 65, 70, 75, 78, 85, 88]
}
df = pd.DataFrame(data)
df
```

```python
y = df['FinalMarks']
y
```

```python
y_predict_baseline = np.repeat(y.mean(),len(y))
y_predict_baseline
```

```python
mae = mean_absolute_error(y,y_predict_baseline)
print(round(mae,2))
```

```python
mse = mean_squared_error(y,y_predict_baseline)
print(round(mse,2))
```

```python
rmse = np.sqrt(mse)
print(round(rmse,2))
```

```python
# syntax
# r2=r2_score(actual_value,pedicted value)
# r-squared score is to check goodness to fit, it,s value lise 0 to 1, minus is very bad, its good as closer to 1
r2=r2_score(y,y_predict_baseline)
print('R-Squared Score is: ',r2)
```

```python
print(f'Baseline MAE={mae:.2f}, RMSE={rmse:.2f}, R-Squared={r2:.2f}')
```

```python
# Clean the data
print(df.isnull().sum())
```

```python
df = df.fillna(df.mean())
df
```

```python
# Data prep, find input, output
X = df[['StudyHours','Attendance','PrevScore']]
y = df['FinalMarks']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
# Model Creation
model = LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
y_predict = model.predict(X_test)
print(y_predict)
```

```python
mae = mean_absolute_error(y_test,y_predict)
print(mae)
```

```python
mse = mean_squared_error(y_test,y_predict)
print(mse)
```

```python
rmse = np.sqrt(mse)
print(rmse)
```

```python
r2 = r2_score(y_test,y_predict)
print(r2)
```
