```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score,r2_score
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
data = {
    'Area':[850,900,1000,1100,1200,1500,1600,1800,2000],
    'Bedroom':[1,2,2,2,3,3,2,4,4],
    'Age': [1,1,3,2,1,2,2,1,2],
    'PriceIn100K':[5,6,7,7,7,9,9,11,None]
}
df = pd.DataFrame(data)
df
```

```python
y = df['PriceIn100K']
y
```

```python
df_filled = df.fillna({
    'PriceIn100K': df['PriceIn100K'].mean()
})
print(df_filled)
```

```python
y = df_filled['PriceIn100K']
y
```

```python
y_predict_baseline = np.repeat(y.mean(),len(y))
y_predict_baseline
```

```python
mae = mean_absolute_error(y,y_predict_baseline)
print(mae)
```

```python
mse = mean_squared_error(y, y_predict_baseline)
print(mse)
```

```python
rmse = np.sqrt(mse)
print(rmse)
```

```python
r2=r2_score(y,y_predict_baseline)
print('R-Squared Score is: ',r2)
```

```python
print(df.isnull().sum())
```

```python
df = df.fillna(df.mean())
df
```

```python
# Data prep, find input, output
X = df[['Area','Bedroom','Age']]
y = df['PriceIn100K']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.7,random_state=42)
```

```python
model = LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
y_predict = model.predict(X_test)
print(y_predict)
```

```python
mae = mean_absolute_error(y_test,y_predict)
print(mae)
```

```python
mse = mean_squared_error(y_test,y_predict)
print(mse)
```

```python
rmse = np.sqrt(mse)
print(rmse)
```

```python
r2 = r2_score(y_test,y_predict)
print(r2)
```
