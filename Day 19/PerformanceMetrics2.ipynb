```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score,r2_score
from sklearn.linear_model import LinearRegression
from math import sqrt
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {
    'Experience':[1,2,3,4,5],
    'Salary':[3500,4000,4300,4800,5200]
}

df = pd.DataFrame(data)
df
```

```python
# Splitting the data
X = df[['Experience']] # input experience
y = df['Salary']       #output
```

```python
# Create the model
model = LinearRegression()
```

```python
# Train the model
model.fit(X,y)
```

```python
#Prediction : predict on same data
predict_salary = model.predict(X)
print(predict_salary)
```

```python
mae = mean_absolute_error(y,predict_salary)
```

```python
print(mae)
```

```python
mse=mean_squared_error(y,predict_salary)
print(mse)
```

```python
mse=mean_squared_error(y,predict_salary)
rmse = sqrt(mse)
print(rmse)
```

```python
# syntax
# r2=r2_score(actual_value,pedicted value)
# r-squared score is to check goodness to fit, it,s value lise 0 to 1, minus is very bad, its good as closer to 1
r2=r2_score(y,predict_salary)
print('R-Squared Score is: ',r2)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(X,y,color='blue', label='Actual Salary',s=100)
plt.plot(X,predict_salary, color='green', linewidth=2, label='Predict Salary')
for i in range(len(X)):
   plt.plot([X.iloc[i, 0], X.iloc[i, 0]], [y.iloc[i], predict_salary[i]], color='yellow', linestyle='--', linewidth=2)
plt.title('Actual Salary vs Predicted Salary')
plt.xlabel('Experience in years')
plt.ylabel('Salary of person')
plt.legend()
plt.show()
```

# Exercise: Predict sales revenue based on advertising budget
## Train a linear regression model based on the above data
## Compute MAE,MSE,RSME and visualize the esult with label plot

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
from sklearn.linear_model import LinearRegression
from math import sqrt
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {
    'advertising_spend':[5,10,15,20,25,30,35,40,45,50,55,60],
    'revenue_gain':[350,370,400,430,450,480,500,520,550,580,600,630]
}
df = pd.DataFrame(data)
df
```

```python
#Splitting the data
X = df[['advertising_spend']]
y = df['revenue_gain']
```

```python
# Model creation
model = LinearRegression()
```

```python
#Train the model
model.fit(X,y)
```

```python
# Prediction
predicted_revenue = model.predict(X)
print(predicted_revenue)
```

```python
mae = mean_absolute_error(y,predicted_revenue)
print(mae)
```

```python
mse = mean_squared_error(y,predicted_revenue)
print(mse)
```

```python
mse=mean_squared_error(y,predicted_revenue)
rmse = sqrt(mse)
print(rmse)
```

```python
# syntax
# r2=r2_score(actual_value,pedicted value)
r2=r2_score(y,predicted_revenue)
print('R-Squared Score is: ', r2)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(X,y,color='blue', label='Actual Revenue',s=100)
plt.plot(X,predicted_revenue, color='green', linewidth=2, label='Predict Revenue')
for i in range(len(X)):
   plt.plot([X.iloc[i, 0], X.iloc[i, 0]], [y.iloc[i], predicted_revenue[i]], color='yellow', linestyle='--', linewidth=2)
plt.title('Advertising Spend vs Revenue Gain')
plt.xlabel('Advertising Spend')
plt.ylabel('Revenue Gain')
plt.legend()
plt.show()
```
