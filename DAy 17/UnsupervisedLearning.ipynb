# K Means Clustring Example
## K-means comes under unsupervised Machine Learning Technique
## Terms Cluster will be used
## Centroid: nearest mean cluster
## Cluster: will partition the dataset into K distinct cluster

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
X = np.array([
    [7.5, 8.0],
    [9.0, 7.0],
    [10.5, 6.5],
    [6.0, 9.0],
    [13.0, 5.5],
    [8.0, 8.5],
    [14.0, 4.5],
    [9.5, 7.0],
    [7.0, 9.0],
    [12.0, 6.0]
])
data = pd.DataFrame(X,columns=['EntryTime','Duration'])
data
```

```python
# Crate the model
model = KMeans(n_clusters=3, random_state=0)
```

```python
data['Cluster'] = model.fit_predict(data[['EntryTime','Duration']])
centers = model.cluster_centers_
```

```python
data['Cluster']
```

```python
plt.figure(figsize=(7,5))
plt.scatter(data['EntryTime'], data['Duration'], c=data['Cluster'], s=100)
plt.scatter(centers[:,0], centers[:,1], marker='X', s=200,c='red', label='Controid')
plt.xlabel('Entry Time (Hour of Day)')
plt.ylabel('Duration in Office (Hours)')
plt.title('Working Hour Pattern Using K-Means')
plt.legend()
plt.grid(True)
plt.show()
```

# Lab: Customer Segmentation using K-Means
## Goal: Group customers by Annual Income and Spending Score.
## 1. Download customer.csv from https://github.com/salman1256/aimltraining/tree/main/Day-17(II)
## 2. Load data from customers.csv
## 3. Visualize the data
## 4. Cluster
## 5. Interpret each segment like 'High income / high spender' etc.
## 6. Create Scatter plot for the same

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
data = pd.read_csv('customers.csv')
data.head(3)
```

```python
# CLuster
X = data[['AnnualIncome','SpendingScore']]
kmeans = KMeans(n_clusters=4, random_state=0)
labels = kmeans.fit_predict(X)
data['Cluster'] = labels
centers = kmeans.cluster_centers_
data.head()
```

```python
plt.figure(figsize=(6,5))
plt.scatter(data['AnnualIncome'], data['SpendingScore'], c=data['Cluster'], s=100)
plt.scatter(centers[:,0], centers[:,1], marker='X', s=200,c='red', label='Centroid')
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.title('Spending Pattern Using K-Means')
plt.legend()
plt.grid(True)
plt.show()
```
